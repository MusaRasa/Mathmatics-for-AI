# Scaler multiplaction

```python
### basic code  by opeatrions (adding, subtracting, multiplaction) is straighforward.
v1 = np.array([2,3,4])
v2 = np.array([6,4,3])
v3 = 4*v1-3*v2
4*v1,'...............',3*v2,'.......',v3
```

### Vector dot product
The dot product is a single number that provide information about the relationship between two vectors.

```python
dot1 = np.array([2,3,4,5])
dot2 = np.array([6,5,4,3])
result = np.dot(dot1,dot2)
print(result)
```
