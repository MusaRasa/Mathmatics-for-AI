### Trangular matrix
* is a half-way between a diagonal and full matrices.
* IT is devided in two part Upper trangular and lower trangular, to get more information you can reject to linear algebra book
* `tril` extract the lower trangular
* `triu` extract the upper trangular

```python
T = np.random.randn(4,4)
lower = np.tril(T)
upper = np.triu(T)
print(lower)
```

# shifting a matrix
* We can combine matrix addition with scalar-matrix multiplicationto apply a procedure called "shifting" a matrix.
* To "shift" a matrixmeans to add to the matrix a multiple of the identity matrix.
* This is the shiftting a matrix fourmula: `A' = A + λI`

```python
lemda = 0.2
I = np.eye(4)
shifittingMatrix = np.random.randn(4,4)
sheftting = shifittingMatrix + I*lemda
I
```

```python
print(shifittingMatrix)
```

```python
print(sheftting)
```

### Trace
* The trace is an operation that produces a single number from a squre matrix.
*   it is indecated to tr(A).
*   The trace operation has two applications in machine learning: It is used to compute the Frobenius norm of a matrix (a measure of the
 magnitude of a matrix) and it is used to measure the "distance" between two matrices.
* The trace is defined only for square matrices

```python
Matrix = np.random.randn(4,4)
traces = np.trace(Matrix)
Matrix
```

```python
print(traces)
```
