## # Sigmoid function formula
* * * We use the S = 1/(1+exp(-x)) formual in Machine learning.
* Why we use it?
* For classification model to 0 and 1
* * * * if x >= 0.5 we can boundred our data to one class
* * * * if x < 0.5 we can boundred our data to zero class

```python
import torch as tr
import numpy as np
import matplotlib.pyplot as plt
```

```python
a = tr.arange(-10,10,1,dtype = tr.float32)
a
```

```python
def sigmoid(x):
    return 1 / (1+tr.exp(-x))
```

```python
plt.plot(tr.sigmoid(a));
plt.grid()
plt.show()
```
